###### A quick note about using an entire column of a dataframe

In Lab 3, we learned about the <code>loc</code> and <code>iloc</code> functions to index a dataframe. You can also index an entire column of a dataframe in a more streamlined way by using square brackets around the column name. The main difference is that using <code>loc</code> and <code>iloc</code>, you can also specify which rows you are interested in. But if you put <code>:</code> for the row index (meaning that you want all of the rows of a specific column) then all three lines below are equivalent: 


```python
#Extract the entire allDate column using iloc:
date_column=water_df.iloc[:,0]

#Extract the entire allDate column using loc:
date_column=water_df.loc[:,'allDate']

#Extract the entire allDate column using square brackets instead of loc or iloc
date_column=water_df['allDate']
```

###### Convert Column to DateTime Type

If you look at values of the allDate column as shown from the first section, we can clearly see that it is indicating a date. However, according to Python, the data type (as shown above as Dtype) of the allDate column is an object. So, in the "brain" of the computer, it interprets the values in the allDate column as just numbers and slashes, not actual dates. This will be important as we begin plotting with this data. We fix this issue by using the <code>to_datetime</code> function from the pandas package. Below, we replace the allDate column with itself, except the values have been converted to the datetimes. 


```python
#Convert allDate column to datetimes
water_df['allDate']=pd.to_datetime(water_df['allDate'])

#Check the info again to make sure it worked
water_df.info()
```

Success! We can now see that there the column allDate has the type of datetime instead of an object. We are now ready to start making a plot.

##### Creating a Plot
**Figure Handle**

The first thing to do when creating any plot is to create a new figure. This is done by using the <code>plt.figure()</code> function. This creates a figure, or a "box" in which the figure is created. It's also useful to assign the figure a variable name to refer to the figure later when we print the figure to a pdf or image.

**Create the Graphic**

Next, use a function that creates the plot we want. Here, it's shown using the "plot" function, but this could also be the boxplot or histogram function as we will see later.

The plot function plots two arrays against each other. The first array entered to the function is on the x axis, and the second is on the y axis. See below for the actual code example. In this case, the dates are on the x axis and the corresponding water consumption is on the y axis. In general though, when using the <code>plot</code> function, it works like this: <code>plt.plot(X_Variable,Y Variable)</code>.

**Adjust some Display Settings**

If the x axis variable is a datetime, then it helps to specify some additional options to make the x axis look better. Note that this ONLY applies when the x axis is a datetime, you do NOT need this when the x axis is not a datetime. 

**Add Labels**

All graphs should have x and y axis labels, and also sometimes a title. They can be added by using the <code>xlabel</code> , <code>ylabel</code>, and <code>title</code>. See below for an example.

Note that the <code>xlabel</code>, <code>ylabel</code>, and <code>title</code> functions can be used no matter what the plot looks like. This is to say that below when we learn the functions to make a histogram and box plot, <code>xlabel</code>, <code>ylabel</code>, and <code>title</code> can, and should still be used. 


```python
#Create a New Figure
fig1=plt.figure()

#Create the plot with the date on the X axis and water consumption on the Y axis
plt.plot(water_df['allDate'],water_df['wF'])

#Format X axis. We'll use this command only when we have dates on the X axis.
#It autoformats the dates to clean it up and make the axis look better
plt.gcf().autofmt_xdate()

#Label the axes and add a title. Always remember to label your axes and include units when applicable!
plt.xlabel('Date')
plt.ylabel('Water Consumption (meters cubed)')
plt.title('Water Consumption in Jester Hall')
```

The default of the plot function is to connect each data point with a blue line. We can vary the display style of the plot with many options, including color, line thickness, and many other options. Below, it is shown how to have each x/y pair shown as a dot instead of the line connecting them. 

##### Creatting a Scatter Plot
The code in the cell below is almost identitcal to the code in the cell above. The only difference is the additional inputs to the plot function which specify that there should be a dot at each data point, <code>marker='o'</code>, and that there should not be a line drawn between points,<code>ls=' '</code>. (Note that ls refers to line style, and here we are telling the line style to be blank).

To learn more about the different display options of using the plot function, look at the documentation for this function:

https://matplotlib.org/3.3.1/api/_as_gen/matplotlib.pyplot.plot.html


```python
#Create a New Figure
fig2=plt.figure()

#Create the plot with the date on the X axis and water consumption on the Y axis
#Here the additional options are added to change the way the data is displayed
plt.plot(water_df['allDate'],water_df['wF'],marker='o',ls='')

#Format X axis
plt.gcf().autofmt_xdate()

#Add labels and title
plt.xlabel('Date')
plt.ylabel('Water Consumption (meters cubed)')
plt.title('Water Consumption in Jester Hall')

fig2.savefig('../figures/jes_water_scatter.pdf', dpi=200)
```

##### Boxplot

To make a boxplot, use the <code>boxplot</code> function. Note that the matplotlib boxplot function requires that the input be a numpy array, not a dataframe, and not have any nan values. First, the column of the dataframe that we are interested in needs to be converted to an array. Second, any nan values need to be removed. Use the <code>to_numpy()</code> to convert from a dataframe to an array, and then <code>water_a=water_a[~np.isnan(water_a)]</code> to remove any nan values (we don't have any nan values this week, but you might in the future). Note that in this line, where it currently says <code>water_a</code> would be whatever name of the variable you're working with. We will cover the mechanics of how this works next week.


```python
#Convert the wF column of the dataframe to an array
water_a=water_df['wF'].to_numpy()

#Create a figure
fig3=plt.figure()

#Create the boxplot
plt.boxplot(water_a)

#Add labels and title
plt.xlabel('Observations')
plt.ylabel('Water Consumption (meters cubed)')
plt.title('Jester Hall Water Consumption Boxplot')
```

##### Histogram

To make a histogram, use the <code>hist</code> function. Again, the input to the function needs to be a numpy array without nan values. The code is repeated from above to show that process. But if you're doing this on your own, you wouldn't need to convert the data to an array and then remove nans twice.


```python
#Convert the wF column of the dataframe to an array
water_a=water_df['wF'].to_numpy()

#Create a figure
fig4=plt.figure()

#Create the histogram
plt.hist(water_a)

plt.xlabel('Water Consumption (meters cubed)')
plt.ylabel('Number of Observations')
plt.title('Water Consumption in Jester Hall')
```

##### Histogram- change the number of bins

The <code>hist</code> automatically determines the number of bins in the histogram. But note that you can also manually specify the number of bins directly. See below.


```python
#Convert the wF column of the dataframe to an array
water_a=water_df['wF'].to_numpy()

#Create a figure
fig5=plt.figure()

#Create the histogram and specify 20 bins
plt.hist(water_a,20)

plt.xlabel('Water Consumption (meters cubed)')
plt.ylabel('Number of Observations')
plt.title('Water Consumption in Jester Hall')
```

##### Controlling Font Size

If we don't specify the font size for the labels and title, the default sizes will be used. We can change and specify the font size of the axis labels and title. Enter the font size when creating the labels and title. This format can be used on all the plot types covered today. To demonstrate, let's recreate the last histogram with bigger labels and title:


```python
#Create a figure
fig6=plt.figure()

#Create the histogram and specify 20 bins
plt.hist(water_a,20)

#Add labels and title with specified font sizes
plt.xlabel('Water Consumption (meters cubed)',fontsize=12)
plt.ylabel('Number of Observations',fontsize=12)
plt.title('Water Consumption in Jester Hall',fontsize=18)
```

##### Explanatory Data Analysis

You are already familiar with the mean, meadian, trimmed mean, variance, standard deviation, and interquartile range of a data set from lecture. See below for how to calcuate these values in python. Note that the <code>trim_mean</code> and <code>iqr</code> functions are from the scipy package, not numpy. 

#### Note about NaN values: 
For the numpy functions, if there are NaNs in your data, use the np.nan__ version of the function to skip over NaNs. Note that the two SciPy functions do not work with NaNs at all, so your data needs to have all NaNs removed first.


```python
#Numpy functions:

#mean
#If there are NaNs in your data, use np.nanmean
water_mean=np.mean(water_df['wF'])
print('The mean is ', water_mean)

#Median
#If there are NaNs in your data, use np.nanmedian
water_median=np.median(water_df['wF'])
print('The median is ', water_median)

#Variance
#If there are NaNs in your data, use np.nanvar
water_var=np.var(water_df['wF'])
print('The variance is ', water_var)

#Standard Deviation
#If there are NaNs in your data, use np.nanstd
water_std=np.std(water_df['wF'])
print('The standard deviation is ', water_std)


#SciPy Functions - remove NaNs first:
#Trimmed Mean with 10% of the data trimmed on the top and bottom
water_trimmed_mean=sc.trim_mean(water_df['wF'],.1)
print('The trimmed mean is ', water_trimmed_mean)

#Interquartile Range
water_iqr=sc.iqr(water_df['wF'])
print('The IQR is ', water_iqr)
```
