## 1. Types, comments, and variables

Integer values have type `int`, decimals are `float`, text is `str`. Use `#` for a single-line comment.


```python
print(1 / 2)              # division always returns a float
print('Hello World')

x = 1 / 2                 # assignment does not print
print(x)
print(type(x))

x = 5 * 5                 # overwritten with an int
print(x, type(x))

x = 'I love coding'       # now a string
print(x, type(x))
```

Python follows standard operator precedence (parentheses, exponents, multiplication/division, addition/subtraction).


```python
print(100 / 100 * 100)
print(100 / (100 * 100))
result = (2 + 3) * (5 - 2) / 4 + 6**2 - 1
print(result)
```

## 2. Worked example: NumPy


```python
x = np.linspace(0, 30, 31)  # 31 values from 0 to 30 inclusive
y = np.linspace(0, 1, 100)
print(x[:5], '...', x[-3:])

z = np.sqrt(49)
z = np.deg2rad(z)
z = np.round(z, 3)
print('ln(z) =', np.log(z))
print('log10(1e5) =', np.log10(100000))
print('pi =', np.pi)
print(np.sin(np.pi / 2), np.cos(np.pi), np.arcsin(0.5))
```

```python
z = np.arange(50)
print('sum', np.sum(z), 'max', np.max(z), 'min', np.min(z), 'mean', np.mean(z))
```
