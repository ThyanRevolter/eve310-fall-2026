##### Distinctions between dataframes and arrays
Properties shared by arrays and dataframes:
1. Stores data in a matrix (rows and columns)
2. Can conduct mathematical operations
3. Ability to convert between dataframes and arrays

Properties unique to an array include:
1. Can only contain numbers
2. Does not have a description of data embedded in the python variable
3. Based on numpy package
4. Index based on column and row numbers

Properties unique to a dataframe include:
1. Can use numbers or strings but each column must be the same type
2. Column and row names are embedded in the python variable
3. Based on pandas package
4. Index based on column and row numbers, or on the column and row names


#### Size and Shape
When working with dataframes, it's useful to know how big your dataframe is (how many rows and columns it has). To do that, we use two different functions: <code>size</code> and <code>shape</code>. <code>size</code> returns how many total values are stored within the dataframe. <code>shape</code> returns the number of rows and columns within the dataframe. Intuitively, the value of <code>size</code> is the product of the number of rows and columns returned by <code>shape</code>.

Here, the <code>size</code> and <code>shape</code> functions are used on dataframes, but they also work on arrays to. The <code>size</code> and <code>shape</code> functions do not work with lists.


```python
#size
siz=water_df.size
print('The size of the dataframe is', siz)
#shape
[row,col]=water_df.shape
print('The number of rows in the dataframe is',row)
print('The number of columns in the dataframe is',col)


print('The number of rows multiplied by the number of columns is',row*col)
```

###### Renaming, Removing and Adding columns to a dataframe

When working with datasets, it us useful to be able to rename a column, remove a column, or add a column to a dataframe. 

###### Rename a column
To rename a column, use the <code>rename</code> method which renames a column in a dataframe. See below for how the function is used, where the existing column name, wF, is changed to be Water Usage.

You will notice that the variable <code>water_df</code> is being re-assigned its value based on the modified <code>water_df</code> dataframe. This is a common practice in python, where a modified variable will overwrite itself.


```python
#Rename the column
water_df = water_df.rename(columns = {'wF':'Water Usage'})

#Show the first 3 rows to see the different column name
water_df.head(3)
```

###### Remove a column
To remove a column, use the <code>drop</code> method which drops a column in a dataframe. See below for how the function is used, where the prevWeek column is removed from the dataframe.

Again, the variable <code>water_df</code> is being re-assigned its value based on the modified <code>water_df</code> dataframe. 


```python
#Removing a column
water_df=water_df.drop(['prevWeek'],axis=1)

#Taking a look at what we did. See! No prevWeek column anymore!
water_df.head(3)
```

###### Adding a column
To add a column, assign a new column values by using the new column name as a string inside brackets referencing the dataframe. See below for an example of adding a column named 'Twos' to the dataframe, in which each row has the value of 2. Note that when adding values to a new column, the values of that new column should be an array or list of 1 column, and the number of rows that already exist in the dataframe.


```python
#Add a column named Twos with the value of 2 in each row
water_df['Twos'] = np.ones(len(water_df)) * 2

#Taking a look at what we did.
water_df.head(3)
```

###### Using subsets of a dataframe

Similar to arrays and lists, dataframes can be indexed to only specify a subset of a dataframe instead of the entire dataframe. There are two different ways to do this: 

1. Reference a specific subset of rows and columns by using integer indexing, and using the <code>iloc</code> function. This uses integer indexing in the same way we previously learned to index an array or list. See below for an example. <code>iloc</code> stands for "integer location."
2. Reference a specific subset of rows and columns by using the row and column names, and using the <code>loc</code> function. This works similar <code>iloc</code>, but instead of using integers to index dataframe, the row and column names are used instead. See below for an example. <code>loc</code> stands for "location." 


```python
#Print the first value of the first column of water_df
print(water_df.iloc[0,0])

#Print the first value of the column named allDate
print(water_df.loc[0,'allDate'])


#Two different ways of doing the same thing!
```
