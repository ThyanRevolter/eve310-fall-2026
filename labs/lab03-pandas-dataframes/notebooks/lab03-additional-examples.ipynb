# Additional Examples of Adding Columns to Data Frames


Make sure you run the block of code below which will import the packages. The rest of the code won't work if you don't run this code block first.


```python
#import packages
import pandas as pd
import numpy as np
```

Create a simple dataframe with two columns

Here, we are using the <code>pd.DataFrame</code> function to create a new dataframe from scratch. In the other examples in the tutorial and in class, the dataframe is created when the csv file is imported using the <code>pd.read_csv</code> function.


```python
#Create and show a simple dataframe
my_df=pd.DataFrame(columns=['A','B'], data=np.array([[4,6],[9,1],[4,7],[9,3]]))
my_df.head(4)
```

Add a new column called C with the values 6,3,2,1


```python
new_variable=np.array([6,3,2,1])
my_df['C']=new_variable

my_df.head(4)
```

Add a new column D that has the same values as column A


```python
my_df['D']=my_df.loc[:,'A']
my_df.head(4)
```

Add a new column E which is half the value of column B


```python
col_b=my_df.loc[:,'B']
my_df['E']=col_b*.5
my_df.head(4)
```

Add a new column F which has the value of 8 for each row


```python
#putting (4,1) inside the ones function tells it that you want an array with 4 rows (because the dataframe has 4 rows)
#and 1 column. 

#Each element of that array will have the value of 1
#We then myltiply by 8 to get each value to equal 8 since 8*1=8
new_col=np.ones((4,1))*8
my_df['F']=new_col
my_df.head()
```
