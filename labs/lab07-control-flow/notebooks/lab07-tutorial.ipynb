**Logical Operators**

Before getting to if statement, first the idea of logical operators is introduced. Usually when we are using variables in python, we are assigning values to variables. For example, <code>a=2</code> is telling the computer "The variable a has the value of 2!"  However, we can also ask questions of python using logical operators. The idea of a logical operator is that it asks some logical question of two values, and returns a value of *true* or *false* depending on the truth value of the logical statement.

Below each of these logical operators is demonstrated.

Again, be careful of the difference between <code>x = y</code>, which assigns the value of y to the value of x, and  <code>x == y</code> which asks if the values of x and y are equal.


```python
#Assign values to a and b so we can use them in logical statements

a = 5
b = 4

#Equal to
print(a == b)

#Greater than
print(a > b)

#Greater than or equal to
print(a >= b)

#less than
print(a < b)

#less than or equal to
print(a <= b)

#Not equal to
print(a != b)
```

**If Statements**

If statements evaluate a logical operator statement. If the statement is true, then the code within the if statement executes. If not, then the code within the if statement is skipped over.

Code that is indented is within the if statement. Code that is not indented is not included in the if statement. An easy way to indent code is to use the tab key.

Here we are just demonstrating a simple if statement, but the code within the if statement an be as long as you want it to be. You also could have if statements within if statements.

You can change the values of a and b to see what happens under different scenarios.


```python
#If Statements
a = 3
b = 2

#If A is greater than B, print something to the screen
if (a > b):
    #Code indented here is included in the if statement
    print('The code in the if statement has run')

#This code here is not included in the if statement
print('This is not in the if statement so it will run regardless of the values of a and b')
```

**If Else Statements**

Suppose we have some code that we want to run if a condition is true. In that case a "normal" if statement will work great. However, in other cases, we will have some code we want to run if the condition is true, and other code that we want to run if the condition is false. Here, we demonstrate this by using an else statement. In an if-else statement, the code indented within the if statement will run if the logical condition is true, and the code within the else statement will run if the logical condition is false.


```python
a = 3
b = 2

if (b > a):
    #This executes if the conditional statement is true
    print('B is greater than A')
    
else:    #This executes if the conditional statement is false
    print('B is less than A')
```

**Precursor to For Loops**

A for loop is a useful technique to repeat a task many times. A for loop works by altering the value of a variable, and repeating that code multiple times. This concept is best explained by example.

Suppose we create an array, called my_array, as <code>my_array = np.array([6,5,8,3])</code>. The task to be completed is to print each value of my_array to the screen. Below we show two different ways that this task can be done. With a small array of just 4 values, it's not hard to do just by typing each value.


```python
#For Loops

#Let's make an array
my_array = np.array([6,5,8,3])

#Suppose I wanted to print each element of the array d. I could do this

print('By manually typing in the indicies:')
print(my_array[0])
print(my_array[1])
print(my_array[2])
print(my_array[3])

#I could also do this
print('By changing the value of i, used as the index:')
i = 0
print(my_array[i])
i = 1
print(my_array[i])
i = 2
print(my_array[i])
i = 3
print(my_array[i])
```

**For Loops**

Notice how by using a variable, i, and changing the value of it, the task is also completed. A for loop changes the value of an index automatically, and repeats a task using different values of that index in each loop.

There are 5 main components of a for loop (as shown below):
for _(variable to changed)_ in _(the possible values of that variable)_:

In any for loop, it starts with the word <code>for</code>, which tells python that a for loop is being initiated.

Next, the variable to changed. This is the variable to be used inside of the for loop which will have its value changed on each iteration.

Next, is the word "in"   This will be the same in any for loop you will ever do.

Next is a variable which represents the possible values of the variable to be changed. This variable is usually an array. Sometimes the range function is used. In the generic example range(a,b), the value of i would take each integer value from a to 1-b.

Then, it's all finished by a :

In the example below, the value of i will take the values of 0,1,2,3 within the loop.

Similar to if statements, the code to be run within the for loop needs to be indented.

Imagine how powerful a for loop can be! If there were 10000 values of my_array and we wanted to print them all, we could copy-paste the print command 10000 times, or use a for loop with 2 lines of code.


```python
#I could also do this (Same as in the above code block)
print('By changing the value of i, used as the index:')
i = 0
print(my_array[i])
i = 1
print(my_array[i])
i = 2
print(my_array[i])
i = 3
print(my_array[i])

#Also using a for loop.
print('By using a for loop:')
for i in range(0,4):
    print(my_array[i])
```
