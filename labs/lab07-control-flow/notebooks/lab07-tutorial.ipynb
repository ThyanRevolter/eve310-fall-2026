**The for loop**

A for loop changes the value of a variable automatically and runs the indented code once for each value. Its first line has five parts:

`for` *(the variable that changes)* `in` *(the values it takes)* `:`

1. `for` tells Python a loop starts here
2. the variable, usually called `i`, that gets a new value on every pass
3. the word `in`, the same in every for loop
4. the values `i` takes, one per pass: here `range(0, 4)` gives 0, 1, 2, 3
5. a colon `:` to end the line

As with an if statement, the code that belongs to the loop is **indented**. The loop below prints the four values with two lines of code. If `my_array` had 10,000 values, it would still be two lines.

**Type it:** the loop line, `for i in range(0, 4):`, and under it an indented print of `my_array[i]`.

**`range`**

Writing `range(len(my_array))` instead of `range(0, 4)` makes the loop work for an array of any length. To see the values a `range` holds, wrap it in `list(...)`.

## 4. A loop and an if statement together

The task: find the **maximum** of `my_array` without `np.max`. The idea is to keep a variable `max_val` that holds the largest value seen so far:

1. Start `max_val` at the **first** value of the array, `my_array[0]`.
2. Loop over every index `i`.
3. If `my_array[i]` is greater than `max_val`, it is the new largest value seen so far, so store it in `max_val`. Otherwise do nothing.

**Common mistakes**

| What you see | What went wrong |
| --- | --- |
| `SyntaxError: expected ':'` | The `if`, `elif`, `else` or `for` line is missing its colon |
| `IndentationError` | The line after a colon is not indented, or the lines inside a block are indented by different amounts |
| The loop body runs once, after the loop | The line is not indented, so it is outside the loop |
| `SyntaxError: invalid syntax` at `if a = b` | `=` assigns; the question is `==` |
| The last value is never visited | `range(0, n)` stops at `n - 1`. Use `range(len(my_array))` and the last index takes care of itself |
| A count that keeps growing | The accumulator was set to 0 in another cell, so re-running the loop adds to the old count |
| `SyntaxError` on a line with `--` | A placeholder is still there. Every `--` in a cell must be replaced before the cell can run |
