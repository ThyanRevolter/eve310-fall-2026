##### Lists, Arrays, and Indexing

Sometimes it is useful to have individual numbers be variables on their own. But in many cases, that would not be appropriate. For example, consider analyzing water consumption of a building, and you have a spreadsheet of water consumption each day for a few months. That would be about 90 individual values. It would be terribly iniffecient, and many operations you could not even perform, if each of those values were stored as its own variable. That's why there are additional variable types, lists, arrays, and dataframes (and others) which are usefull for working with groups of data. For this, we introduce lists, which are variables which can contain numbers or strings, arrays, which contain only numbers and can be used to perform mathematical operations. Then, we'll introduce indexing, which is a useful technique to refer to certain elements of either a list or array.

The first type of variable we will introduce is a **list**. A list is a variable that stores any type of data (variables or strings) which can be referred to all in one. Here's an example of creating a list:


```python
#Creating a list
my_list = [5,'word',74,'another word']

print(my_list)
```

Here, we can see that each individual element of the list is seperated by a comma, and the whole thing has a square bracket on each end.

Next, here's an example of an **array**. Creating arrays requires the numpy package (introduced above), and uses the array function. The array function takes a list (created by the parts in square brackets below) and turns it into an array (the part in normal parenthesis below). All values in an array must be numbers, no strings.


```python
# Creating an array
my_array = np.array([5,7,4])
print(my_array)
```

Arrays are useful because they allow for mathematical operations to be preformed, unlike lists. Here are just a few of many examples:


```python
# Add to all values in array
my_array_plus_two = my_array+2
print('My array plus two:', my_array_plus_two)

# Multiply all values by 3:
my_array_times_three = my_array*3
print('My array times three:', my_array_times_three)
```

Lists will result in errors when performing mathematical operations.

###### 1 Dimensional Indexing


Making a list or array is nice, but what about if we want to do something with an individual value (or multiple of the values but not the whole list)? For that, we use indexing which allows us to refer to only certain elements, or indiviudal values, of a list.

To do this, we take the name of the list (in this case my_array), and then use square brackets to refer to it's index. an index is a fancy way of referring to a value's "place" in the array. Note that the indexing starts at 0, so the index 0 refers to the first value, 1 refers to the second value, and so on.

Here's some examples of indexing:


```python
# Indexing the array which we already created

# Extracting the first value of my_array, using index 0
first_val = my_array[0]
print(first_val)

# Extracting the second value of my_array, using index 1
second_val = my_array[1]
print(second_val)

# Extracting the first two values of my_array
first_two_vals = my_array[0:2]
print(first_two_vals)
```

###### 2 Dimensional Arrays

So far all of the arrays we have worked with are 1 dimensional, meaning, just one row. However 2 dimensional (2D) arrays are useful to store data that has multiple columns. For example, if working with data about the weather, an array could have the temperature as the first column, and the humidity as the second column, all in one array. Here is a simple example:


```python
# Creating a 2d array
my_array_2D = np.array([[5,7,4],[8,9,2]])
print(my_array_2D)
```

Notice how creating a 1D array differs from creating a 2D array. To create a 1D array, the part inside the parenthesis of the np.array function is: <code>[5,7,4]</code>, just a list. But in a 2D array, the part inthe side the parenthesis is: <code>[[5,7,4],[8,9,2]]</code>. Here, there are square brackets around the whole thing, then individual square brackets around the values that will become each row in the array, with a comma in the middle. Each row in the array is entered as its own list.


###### 2 Dimensional Indexing

As we saw with 1D arrays, 2D arrays can be indexed to refer to specific values, as well. However, instead of indexing with only one value, we use 2 values, to refer to the row and column of the value (or values) we want to access. Again, the values of the first row and first column start at 0. See below for examples:


```python
# Accessing the first row and first column of my_array
first_row_first_col = my_array_2D[0,0]
print('The value of the first row and first column is',first_row_first_col)


# Accessing the first row and second column of my_array
first_row_second_col = my_array_2D[0,1]
print('The value of the first row and second column is',first_row_second_col)


# Accessing the second row and third column of my_array
second_row_third_col = my_array_2D[1,2]
print('The value of the second row and third column is',second_row_third_col)


# Accessing the second row and all columns of my_array
second_row_all_col = my_array_2D[1,:]
print('The value of the second row and all columns is',second_row_all_col)


# Accessing the all rows and the first column of my_array
second_row_all_col = my_array_2D[:,0]
print('The value of all rows and the first column is',second_row_all_col)
```

As the examples above show, the index contains two values, separated by a comma. The first value refers to the *row*, and the second value referes to the *column*. In addition, if we want all of the values from a particular row or column, use a colon (:) instead of a number.


###### Math Operations on Arrays

There are many functions built into the numpy package which conduct math operations on the entire array. Examples of functions include finding the sum, minimum, maximum, mean, and standard deviation of all of the values in an array.

For example, the array <code>my_array</code>, defined as <code>my_array=np.array([3,7,8,1,4,6,8,3,6,5,8,2,6,1,4]</code>, then use <code>np.sum(my_array)</code> to find the sum of that array.


```python
#Define my_array
my_array = np.array([3,7,8,1,4,6,8,3,6,5,8,2,6,1,4])

#Determine some properties of my_array
print('Sum of my_array is',np.sum(my_array))
print('Max of my_array is',np.max(my_array))
print('Min of my_array is',np.min(my_array))
print('Standard deviation of my_array is',np.std(my_array))
```

##### A note about NaN values
NaN (not a number) values are very common in real world data (a sensor could have malfunctioned, there could have been a reporting error, etc.). It's important to know if your data has NaN values, because this can impact which functions to use and how different functions perform.

This week, our data has some NaN values in it. This means we need to be careful about which numpy functions we use. Let's say we want to find the max and min of certain columns. 

The np.min and np.max functions we just learned about do not work with NaNs, so if we use those we will actually get an incorrect answer, or possbily an error message.

Let's try it out with the prevWeek column.


```python
#First, turn the dataframe into an array 
water_array = water_df.to_numpy()

#Extract all rows from the third column to store as a vector
prev_week = water_array[:,2]

#Change datatype of array to float64 
prev_week = prev_week.astype(np.float64)

np_min = np.min(prev_week)
np_max = np.max(prev_week)

print('np.min:',np_min)
print('np.max:',np_max)
```

np.min and np.max produced an outcome, but it's incorrect or produces a nan outcome. 

We can also try the built in Python functions, min and max, but these don't always work when the array starts with a NaN value, as is the case here.


```python
py_min = min(prev_week)
py_max = max(prev_week)

print('py min:',py_min)
print('py max:',py_max)
```

We get an output of nan. 

Instead, we can use the np.nanmin and np.nanmax functions, which are designed to handle NaN values and skip over the NaNs to calculate the correct min and max.


```python
np_nanmin = np.nanmin(prev_week)
np_nanmax = np.nanmax(prev_week)

print('np.nanmin:',np_nanmin)
print('np.nanmax:',np_nanmax)
```

```python
print(np.mean(prev_week))
print(np.nanmean(prev_week))

print(np.median(prev_week))
print(np.nanmedian(prev_week))

print(np.var(prev_week))
print(np.nanvar(prev_week))

print(np.std(prev_week))
print(np.nanstd(prev_week))

import scipy.stats as sc

print(sc.trim_mean(prev_week,.1))
```

In summary: Different functions perform differently when there are NaN values, and we don't always know what's going on in the background. In this case, the np.nanmin and np.nanmax functions are the most robust to handle NaN values, so that is what we should use.
