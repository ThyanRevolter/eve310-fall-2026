## 1. Vectors in Python

An array made from one list is 1-D. It has 3 elements and shape `(3,)`, and it is neither a row vector nor a column vector.

```python
import numpy as np
```

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
print(x)
print(y)
```

`.reshape(-1, 1)` forces the dimensions: 1 column, and the `-1` lets NumPy work out the number of rows. The result is a column vector with shape `(3, 1)`, which prints with one number per line.

```python
x = np.array([1,2,3]).reshape(-1,1)
y = np.array([4,5,6]).reshape(-1,1)
print(x)
print(y)
```

## 2. Addition

Two arrays with the same dimensions add element by element.

```python
z = x + y
print(z)
```

## 4. Combining operations

This cell runs without an error, but it returns a 3 × 3 matrix, not the 3 × 1 vector you get by hand. `a*x` is a column with shape `(3, 1)` and `np.array([4,0,6])` is 1-D with shape `(3,)`. NumPy stretches the two to a common shape (broadcasting) and gives no warning.

```python
print(a*x - np.array([4,0,6]))
```

`a*x` on its own is the 3 × 1 column you expect, so the second vector is the one to fix.

```python
print(a*x)
```

Give both vectors the same dimensions and the subtraction is element by element.

```python
x = np.array([1,2,3]).reshape(-1,1)
a = 10
print(a*x - np.array([4,0,6]).reshape(-1,1))
```

## 5. Vector-vector multiplication

`x` and `y` are both column vectors, 3 × 1.

```python
print(x)
print(y)
```

**Expect an error.** `np.dot` multiplies the way a matrix product does: the number of columns of the first array has to equal the number of rows of the second. Here those are 1 and 3, so this cell stops with `ValueError: shapes (3,1) and (3,1) not aligned`. Go on to the next cell.

```python
var1 = np.dot(x,y)
```

Transpose the first vector. `x.T` is 1 × 3, so `np.dot(x.T, y)` is (1 × 3)(3 × 1), a 1 × 1 result. This is the inner (dot) product: 1·4 + 2·5 + 3·6 = 32.

```python
var1 = np.dot(x.T,y)
print(var1)
```

The element-wise product `np.multiply(x, y)` multiplies matching elements and keeps the 3 × 1 shape.

```python
var2 = np.multiply(x,y)
print(var2)
```

## 6. Matrix-matrix multiplication

The same two matrices, multiplied two ways. `np.dot(A, B)` is the matrix product, rows times columns.

```python
A = np.array([(1,2),(3,4)])
B = np.array([(5,6), (7,8)])
C = np.dot(A,B)
print(C)
```

`np.multiply(A, B)` is element by element. Compare it with the result above.

```python
C = np.multiply(A,B)
print(C)
```

## Takeaways

- Check the dimensions of your arrays before any arithmetic. `print(x.shape)` shows them.
- `.reshape(-1, 1)` turns a 1-D array into a column vector.
- `np.dot` is the inner (matrix) product. `np.multiply` is element by element.
- A cell that runs without an error can still be wrong. Broadcasting returns a result of a different size than you meant, so print the result and check it.
